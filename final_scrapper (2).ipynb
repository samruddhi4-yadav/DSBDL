```python
import requests
from bs4 import BeautifulSoup

url = "https://en.wikipedia.org/wiki/List_of_countries_by_cancer_rate"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

if response.status_code != 200:
    raise Exception(f"Failed to fetch page: {response.status_code}")

soup = BeautifulSoup(response.text, "html.parser")

# Find the first wikitable (main cancer frequency table)
table = soup.find("table", {"class": "wikitable"})

countries_data = []

# Loop through rows (skip header row)
rows = table.find_all("tr")[1:]

for row in rows:
    cols = row.find_all("td")

    if len(cols) >= 2:
        country = cols[0].get_text(strip=True)
        rate = cols[1].get_text(strip=True)

        countries_data.append({
            "country": country,
            "cancer_rate": rate
        })

# Print sample output
for item in countries_data[:10]:
    print(item)
```

```python
cancer_table=soup.find('table',{'class':'wikitable sortable static-row-numbers col1left'})
print(cancer_table)
```

```python
links=cancer_table.find_all('a')
```

```python
countries=[]
for link in links:
    countries.append(link.get('title'))
print(countries)
```
